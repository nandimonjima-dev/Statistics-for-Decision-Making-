```python
#H0: Altona are meeting their avg price (mu = 800000)
#H1 : Altona are not meeting their avg price (mu > 800000)


#set alpha = 5% =0.5

#One-Sample t-Test: because we are comparing a sample mean against a known/claimed population mean ($800,000)
#The population standard deviation is unknow
```

```python
alpha =0.05

altona = df[df['Suburb'] == 'Altona']['Price'].dropna()

t_test, p_val = ttest_1samp(altona, popmean= 800000, alternative = 'greater')

print("T test:", t_test)
print("P value is:", p_val)

if p_val < alpha:

    print("Reject Null Hypothesis") 

else: 

    print("Fail to Reject Null Hypothesis")
```

```python
#"Statistical Conclusion",since p = 0.1537 > α = 0.05, we fail to reject the null hypothesis. There is insufficient statistical evidence to conclude that the mean property price in Altona has increased beyond $800,000."
#"Business Interpretation",although the sample mean ($834,830) appears higher than the claimed $800,000, this difference is not statistically significant — it could simply be due to random sampling variation. The real estate agency's claim of $800,000 cannot be disproved with this data.
```

```python
#H0: No difference in mean prices( mu-summer = mu_winter)
#H1 : There is a difference in mean prices (mu-summer != mu_winter )

#set alpha = 5% =0.5

#This is a two-tailed test because we're checking for any difference (not specifically higher or lower).
```

```python

from scipy import stats


df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

df['Year']  = df['Date'].dt.year
df['Month'] = df['Date'].dt.month


df_2016 = df[df['Year'] == 2016].copy()

def get_season(month):
    if month in [12, 1, 2]:
        return 'Summer'
    elif month in [6, 7, 8]:
        return 'Winter'
    else:
        return 'Other'


df_2016['Season'] = df_2016['Month'].apply(get_season)


summer = df_2016[df_2016['Season'] == 'Summer']['Price'].dropna()
winter = df_2016[df_2016['Season'] == 'Winter']['Price'].dropna()




t_stat, p_val = stats.ttest_ind(summer, winter, equal_var=False)
print("T Test", t_stat)
print("P value is:", p_val)




#Decision
alpha = 0.05
if p_val < alpha:
    print("Reject H₀, Significant difference between Summer & Winter prices")
else:
    print("Fail to Reject H₀, No significant difference")
```

```python
#"Statistical Conclusion",since p = 0.000125 < α = 0.05, we reject the null hypothesis. There is a statistically significant difference between summer and winter property prices in 2016.
```

```python

#"H₀","μ_high ≤ μ_low → No price premium for more bathrooms"
#"H₁","μ_high > μ_low → More bathrooms → Higher price"
```

```python

alpha = 0.05

# Step 1: Split into two groups
bath_data = df[['Bathroom', 'Price']].dropna()

high_bath = bath_data[bath_data['Bathroom'] >  2]['Price']  # > 2 bathrooms
low_bath  = bath_data[bath_data['Bathroom'] <= 2]['Price']  # ≤ 2 bathrooms

print(f"> 2 bathrooms : n={len(high_bath):,}, mean=${high_bath.mean():,.2f}")
print(f"≤ 2 bathrooms : n={len(low_bath):,},  mean=${low_bath.mean():,.2f}")
print(f"Price premium : ${high_bath.mean() - low_bath.mean():,.2f}")

# Step 2: Levene's test — check equal variance
lev_stat, lev_p = stats.levene(high_bath, low_bath)
equal_var = lev_p > 0.05
print(f"Levene p-value : {lev_p:.4f} → equal variances: {equal_var}")

# Step 3: Welch's one-tailed t-test
t_stat, p_two = stats.ttest_ind(high_bath, low_bath, equal_var=equal_var)
p_one = p_two / 2 if t_stat > 0 else 1 - p_two / 2

print(f"t-statistic          : {t_stat:.4f}")
print(f"p-value (one-tailed) : {p_one:.6f}")

# Step 4: Decision
if p_one < alpha:
    print("Reject H₀, 2+ bathrooms command a significant price premium")
else:
    print("Fail to Reject H₀, No significant price premium")
```

```python
#"Statistical Conclusion",Since p ≈ 0.000000 < α = 0.05, we reject the null hypothesis. There is overwhelming statistical evidence that properties with more than 2 bathrooms command a significantly higher price.
#"Business Interpretation",Properties with 2+ bathrooms sell for ~$875,476 more on average. The 95% CI ($815,832 → $935,120) entirely excludes 0, confirming this is a real and massive price premium — not due to chance.
```
